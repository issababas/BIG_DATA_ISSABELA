```python
# ============================================================
# ACTIVIDAD 3 — ESTADÍSTICA DESCRIPTIVA
# ============================================================

# Seleccionamos el año 2020 para realizar el análisis.
# Este año pertenece al periodo histórico de los datos.

dataframe2020 = dataframe[dataframe["Year"] == 2020]

# Seleccionamos únicamente la variable que analizaremos.
# Se eliminan los valores faltantes para trabajar únicamente
# con los datos históricos disponibles.

poblacion2020 = dataframe2020["Annual change in population"].dropna()

# ------------------------------------------------------------
# CÁLCULO DE ESTADÍSTICOS
# ------------------------------------------------------------

media = poblacion2020.mean()
mediana = poblacion2020.median()
varianza = poblacion2020.var()
desviacion_estandar = poblacion2020.std()
minimo = poblacion2020.min()
maximo = poblacion2020.max()

print("Año seleccionado: 2020")
print("Media:", media)
print("Mediana:", mediana)
print("Varianza:", varianza)
print("Desviación estándar:", desviacion_estandar)
print("Mínimo:", minimo)
print("Máximo:", maximo)
```

```python
# ------------------------------------------------------------
# ENTIDADES CON EL VALOR MÍNIMO Y MÁXIMO
# ------------------------------------------------------------

fila_minimo = dataframe2020.loc[
    dataframe2020["Annual change in population"].idxmin()
]

fila_maximo = dataframe2020.loc[
    dataframe2020["Annual change in population"].idxmax()
]

print("Entidad con el valor mínimo:")
print(fila_minimo["Entity"])
print("Valor mínimo:", fila_minimo["Annual change in population"])

print()

print("Entidad con el valor máximo:")
print(fila_maximo["Entity"])
print("Valor máximo:", fila_maximo["Annual change in population"])
```

```python
# ============================================================
# ACTIVIDAD 5 — EXTRACCIONES Y ANÁLISIS ADICIONAL VISTO EN CLASE
# ============================================================

# ------------------------------------------------------------
# 5.1 EXTRACCIÓN DE ENTIDAD Y AÑO
# ------------------------------------------------------------

dataframe_extraccion_entity = dataframe[["Entity", "Year"]]

dataframe_extraccion_entity.head()


# ------------------------------------------------------------
# 5.2 EXTRACCIÓN DEL AÑO 2020
# ------------------------------------------------------------

dataframe2020 = dataframe[dataframe["Year"] == 2020]

dataframe2020.head(3)


# ------------------------------------------------------------
# 5.3 EXTRACCIÓN DE ENTITY Y YEAR PARA 2020
# ------------------------------------------------------------

dataframe_extraccion2020 = dataframe2020[["Entity", "Year"]]

dataframe_extraccion2020.head(3)


# ------------------------------------------------------------
# 5.4 EXTRACCIÓN DIRECTA DE ENTITY Y YEAR PARA 2020
# ------------------------------------------------------------

dataframe_extraccion2020_2 = dataframe[
    dataframe["Year"] == 2020
][["Entity", "Year"]]

dataframe_extraccion2020_2.head(3)


# ------------------------------------------------------------
# 5.5 EXTRACCIÓN DE LOS AÑOS 2005–2020
# ------------------------------------------------------------

dataframe_2005_2020 = dataframe[
    (dataframe["Year"] >= 2005) &
    (dataframe["Year"] <= 2020)
]

print("Dimensiones de la extracción 2005–2020:")
print(dataframe_2005_2020.shape)


# ------------------------------------------------------------
# 5.6 EXTRACCIÓN DE MÉXICO 2005–2020
# ------------------------------------------------------------

dataframe_2005_2020MEX = dataframe[
    (dataframe["Year"] >= 2005) &
    (dataframe["Year"] <= 2020) &
    (dataframe["Entity"] == "Mexico")
]

print("Dimensiones de la extracción de México 2005–2020:")
print(dataframe_2005_2020MEX.shape)

dataframe_2005_2020MEX
```

```python
# ============================================================
# 5.7 NUEVA EXTRACCIÓN — RWANDA 1983–2021
# ============================================================

# Se selecciona Rwanda porque es un país cuyo nombre comienza
# con la letra R.

dataframe_1983_2021_Rwanda = dataframe[
    (dataframe["Year"] >= 1983) &
    (dataframe["Year"] <= 2021) &
    (dataframe["Entity"] == "Rwanda")
]

# Mostramos la extracción completa
dataframe_1983_2021_Rwanda
```

```python
# ============================================================
# 5.8 ESTADÍSTICAS DE RWANDA 1983–2021
# ============================================================

# Seleccionamos la variable "Annual change in population"
# y eliminamos valores faltantes.

rwanda_population = dataframe_1983_2021_Rwanda[
    "Annual change in population"
].dropna()

# Cálculo de estadísticas descriptivas

rwanda_media = rwanda_population.mean()
rwanda_mediana = rwanda_population.median()
rwanda_varianza = rwanda_population.var()
rwanda_desviacion_estandar = rwanda_population.std()
rwanda_minimo = rwanda_population.min()
rwanda_maximo = rwanda_population.max()

print("ESTADÍSTICAS DE RWANDA 1983–2021")
print("--------------------------------")
print("Media:", rwanda_media)
print("Mediana:", rwanda_mediana)
print("Varianza:", rwanda_varianza)
print("Desviación estándar:", rwanda_desviacion_estandar)
print("Mínimo:", rwanda_minimo)
print("Máximo:", rwanda_maximo)
```

```python
# ============================================================
# 5.9 DATAFRAME GENERAL — MÁXIMO Y MÍNIMO
# ============================================================

# Seleccionamos únicamente los registros que contienen
# información histórica en "Annual change in population".

dataframe_historico = dataframe[
    dataframe["Annual change in population"].notna()
]

# Identificamos la fila con el valor mínimo y máximo.

fila_minima_general = dataframe_historico.loc[
    dataframe_historico["Annual change in population"].idxmin()
]

fila_maxima_general = dataframe_historico.loc[
    dataframe_historico["Annual change in population"].idxmax()
]

# Creamos un DataFrame con los resultados.

dataframe_max_min = pd.DataFrame({
    "Tipo": ["Mínimo", "Máximo"],
    "Entity": [
        fila_minima_general["Entity"],
        fila_maxima_general["Entity"]
    ],
    "Year": [
        fila_minima_general["Year"],
        fila_maxima_general["Year"]
    ],
    "Annual change in population": [
        fila_minima_general["Annual change in population"],
        fila_maxima_general["Annual change in population"]
    ]
})

dataframe_max_min
```
