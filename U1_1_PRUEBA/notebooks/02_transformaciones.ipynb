```python
import pandas as pd
import requests
```

```python
# Fetch the data.
df = pd.read_csv("https://ourworldindata.org/grapher/life-expectancy.csv?v=1&csvType=full&useColumnShortNames=true", storage_options = {'User-Agent': 'Our World In Data data fetch/1.0'})

# Fetch the metadata
metadata = requests.get("https://ourworldindata.org/grapher/life-expectancy.metadata.json?v=1&csvType=full&useColumnShortNames=true").json()
```

```python
df.head(5)
```

```python
df.rename(columns={"entity": "Pais", "code": "Codigo", "year": "Anio", "life_expectancy_0": "Esperanza de Vida"}, inplace=True)
```

```python
df.head(3)
```

```python
df.set_index("Pais")
```

```python
df.loc[((df["entity"].str.contains("I", case=False, regex=True)))]
```

```python
df.loc[df["code"].isnull()]
```

```python
df.dropna(subset=["code"], inplace=True)
df.isnull().sum()
```

```python
resultado = (
    df.loc[df["Anio"] == 2020]
    .groupby("Pais")["Esperanza de Vida"]
    .mean()
    .sort_values(ascending=True)
)

resultado.head(10)
```
